```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

students = ["A", "B", "C", "D"]
marks = [70, 85, 60, 90]
plt.figure(figsize=(12, 8))

# 1. Line Plot
plt.subplot(2, 3, 1)
plt.plot(x, y, marker='o')
plt.title("Line Plot")
plt.xlabel("X")
plt.ylabel("Y")

# 2. Scatter Plot
plt.subplot(2, 3, 2)
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")

# 3. Box Plot
plt.subplot(2, 3, 3)
plt.boxplot(marks)
plt.title("Box Plot")

# 4. Bar Chart
plt.subplot(2, 3, 4)
plt.bar(students, marks)
plt.title("Bar Chart")
plt.xlabel("Students")
plt.ylabel("Marks")

# 5. Pie Chart
plt.subplot(2, 3, 5)
plt.pie(marks, labels=students, autopct='%1.1f%%')
plt.title("Pie Chart")

plt.tight_layout()
plt.show()
```
